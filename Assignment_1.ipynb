# Filtering & ConditionalSelection

```python
completed_students = Student_info[Student_info['test preparation course'] == 'completed']
print(completed_students)
```

```python
high_math_students = Student_info[Student_info['math score'] > 70]
print(high_math_students)
```

```python
Student_info.head(5)
```

```python
female_high_scores = Student_info[
    (Student_info['gender'] == 'female') &
    (Student_info['math score'] >= 65) &
    (Student_info['reading score'] >= 65) &
    (Student_info['writing score'] >= 65)
]
print(female_high_scores)
```

```python
count_free_lunch = Student_info[Student_info['lunch'] == 'free/reduced'].shape[0]
print("Number of students with free/reduced lunch:", count_free_lunch)
```

# GroupBy & Aggregation

```python
 Student_info.groupby('gender')['math score'].mean()
```

```python
 Student_info.groupby('parental level of education')[['math score','reading score','writing score']].mean()
```

```python
avg_scores_by_testprep = Student_info.groupby('test preparation course')[['math score','reading score','writing score']].mean()
print(avg_scores_by_testprep)
```

# Visualization

```python
plt.hist(Student_info['math score'], bins=10, color='skyblue', edgecolor='black')
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Number of Students')
plt.show()
```

```python
plt.hist(Student_info['reading score'], bins=10, color='lightgreen', edgecolor='black')
plt.title('Distribution of Reading Scores')
plt.xlabel('Reading Score')
plt.ylabel('Number of Students')
plt.show()
```

```python
plt.hist(Student_info['writing score'], bins=10, color='salmon', edgecolor='black')
plt.title('Distribution of Writing Scores')
plt.xlabel('Writing Score')
plt.ylabel('Number of Students')
plt.show()
```

```python
avg_scores_by_gender = Student_info.groupby('gender')[['math score','reading score','writing score']].mean()
```

```python
avg_scores_by_gender.plot(kind='bar', figsize=(8,6))
plt.title('Average Scores by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Score')
plt.legend(title='Subjects')
plt.show()
```

# INTERPRETATION

Looking at the averages, students who completed the test preparation course generally show higher performance across math, reading, and writing compared to those who did not. The difference suggests that preparation provides a measurable benefit, helping students strengthen their skills before the exam. While the improvement is not absolute for every individual, the overall trend indicates that test preparation is associated with better outcomes. In short, completing the course appears to positively impact student achievement.
